## Raw-to-Silver Mapping

| Bronze Table | Bronze Column | Silver Table | Silver Column | Transformation |
|---|---|---|---|---|
| bronze_complaints | complaint_id | silver_complaints | complaint_id | Retained |
| bronze_complaints | ward_id | silver_complaints | ward_id | Retained |
| bronze_complaints | department_id | silver_complaints | department_id | Retained |
| bronze_complaints | status | silver_complaints | status | TRIM + UPPER |
| bronze_complaints | _ingested_at | silver_complaints | _ingested_at | Retained |
| bronze_complaints | _source_file | silver_complaints | _source_file | Retained |
| bronze_complaints | _ingestion_run_id | silver_complaints | _ingestion_run_id | Retained |
| bronze_complaints | _schema_version | silver_complaints | _schema_version | Retained |
| bronze_departments | department_id | silver_departments | department_id | Retained |
| bronze_departments | department_name | silver_departments | department_name | TRIM |
| bronze_departments | _ingested_at | silver_departments | _ingested_at | Retained |
| bronze_departments | _source_file | silver_departments | _source_file | Retained |
| bronze_departments | _ingestion_run_id | silver_departments | _ingestion_run_id | Retained |
| bronze_departments | _schema_version | silver_departments | _schema_version | Retained |
| bronze_wards | ward_id | silver_wards | ward_id | Retained |
| bronze_wards | ward_name | silver_wards | ward_name | TRIM |
| bronze_wards | _ingested_at | silver_wards | _ingested_at | Retained |
| bronze_wards | _source_file | silver_wards | _source_file | Retained |
| bronze_wards | _ingestion_run_id | silver_wards | _ingestion_run_id | Retained |
| bronze_wards | _schema_version | silver_wards | _schema_version | Retained |

```python
%sql
CREATE OR REPLACE TABLE silver_complaints AS
SELECT
    complaint_id,
    ward_id,
    department_id,
    TRIM(UPPER(status)) AS status,
    _ingested_at,
    _source_file,
    _ingestion_run_id,
    _schema_version
FROM bronze_complaints;
```

```python
%sql
CREATE OR REPLACE TABLE silver_wards AS
SELECT
    ward_id,
    TRIM(ward_name) AS ward_name,
    _ingested_at,
    _source_file,
    _ingestion_run_id,
    _schema_version
FROM bronze_wards;
```

```python
%sql
SELECT *
FROM silver_wards
LIMIT 10;
```
