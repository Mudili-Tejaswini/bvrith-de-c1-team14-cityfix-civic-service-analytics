```python
raw_path = "/Volumes/cityfix/cityfix/cityfix-zenaiz/raw"
```

```python
agencies_df = spark.read.option("header","true").csv(f"{raw_path}/agencies.csv")

boroughs_df = spark.read.option("header","true").csv(f"{raw_path}/boroughs.csv")

categories_df = spark.read.option("header","true").csv(f"{raw_path}/categories.csv")

requests_df = spark.read.option("header","true").csv(f"{raw_path}/requests.csv")

zip_df = spark.read.option("header","true").csv(f"{raw_path}/zip_geography.csv")
```

```python
display(agencies_df)

display(boroughs_df)

display(categories_df)

display(requests_df.limit(20))

display(zip_df)
```

```python
from pyspark.sql.functions import (
    current_timestamp,
    input_file_name,
    lit
)
```

```python
from pyspark.sql.functions import *
```

```python
ingestion_run_id = "week04_run_001"

agencies_bronze = agencies_df \
    .withColumn("source_file", lit("agencies.csv")) \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("ingestion_run_id", lit(ingestion_run_id))
```

```python
boroughs_bronze = boroughs_df \
    .withColumn("source_file", lit("boroughs.csv")) \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("ingestion_run_id", lit(ingestion_run_id))
```

```python
categories_bronze = categories_df \
    .withColumn("source_file", lit("categories.csv")) \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("ingestion_run_id", lit(ingestion_run_id))
```

```python
requests_bronze = requests_df \
    .withColumn("source_file", lit("requests.csv")) \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("ingestion_run_id", lit(ingestion_run_id))
```

```python
zip_bronze = zip_df \
    .withColumn("source_file", lit("zip_geography.csv")) \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("ingestion_run_id", lit(ingestion_run_id))
```

```python
agencies_bronze.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("cityfix.cityfix.bronze_agencies")
```

```python
boroughs_bronze.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("cityfix.cityfix.bronze_boroughs")
```

```python
categories_bronze.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("cityfix.cityfix.bronze_categories")
```

```python
requests_bronze.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("cityfix.cityfix.bronze_requests")
```

```python
zip_bronze.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("cityfix.cityfix.bronze_zip_geography")
```

```python
display(spark.table("cityfix.cityfix.bronze_agencies"))

display(spark.table("cityfix.cityfix.bronze_boroughs"))

display(spark.table("cityfix.cityfix.bronze_categories"))

display(spark.table("cityfix.cityfix.bronze_requests").limit(20))

display(spark.table("cityfix.cityfix.bronze_zip_geography"))
```
